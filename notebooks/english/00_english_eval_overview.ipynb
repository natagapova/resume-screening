# English model evaluation — track overview

**Location:** `notebooks/english/` (this folder)

**Shared implementation:** `english_eval_common.py` (same directory). Model-specific notebooks set `MODEL_RUN_ID` and call `run_english_eval_pipeline(...)`.

**Data (do not redefine here):**

- Splits: `data/processed/english_dataset_v1/` — produced by **`32_english_dataset_alignment.ipynb`**
- Soft slice: `notebooks/results/english_balanced_eval/english_balanced_8class_eval.csv` — produced by **`54_english_balanced_8class_eval.ipynb`**
- Base eval slice: `notebooks/results/english_dataset_alignment/32_english_base_eval_slice.csv`

**Checkpoints:** Resolution order is documented in `notebooks/models/english/checkpoint_resolution.json`. Optional symlinks under `notebooks/models/english/<model_run_id>/` are checked first.

**Outputs (parallel-safe):**

- Tables + metrics: `notebooks/results/english_eval/<model_run_id>/`
- Figures: `figures/english_eval/<model_run_id>/`

Missing CSVs (e.g. soft slice not built yet) are **skipped** with a note in `run_manifest.json`.

**Do not** commit large prediction CSVs if your workflow forbids it; paths are deterministic so you can add them to `.gitignore` locally.


## Paper-facing aggregate (run after per-model evals)

| Notebook | Purpose |
|----------|---------|
| **`english_benchmark_aggregate.ipynb`** (this folder) | Merges every `metrics.json` under `english_eval/` and `english_transfer_eval/` into **`english_benchmark_summary.csv`**, optionally joins Russian-track F1 from **`71`** unified table for Δ(F1_en − F1_ru). **No GPU.** *Not* `notebooks/60_label_smoothing.ipynb` (that is unrelated training). |
| **`59_english_transfer_existing_models.ipynb`** (repo `notebooks/`) | Writes the same metric schema to **`english_transfer_eval/<model_run_id>/`** for explicit **transfer / OOD** framing. |

Export to thesis tables: run **`52_table_export.ipynb`** after the aggregate notebook — generates **T15_english_benchmark_wide** when the summary CSV exists (legacy `60_english_benchmark_summary.csv` is still accepted by **52** if present).
