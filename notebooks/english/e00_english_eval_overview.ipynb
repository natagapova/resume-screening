# e00 — English track overview

**Папка:** `notebooks/english/`

**Общий код:** `e07_english_eval_common.py` — загрузка чекпоинтов, инференс, метрики (`run_english_eval_pipeline`, и т.д.).

**Данные (не переопределять здесь):**

- Сплиты: `data/processed/english_dataset_v1/` ← **`notebooks/32_english_dataset_alignment.ipynb`**
- Soft slice: `notebooks/results/english_balanced_eval/english_balanced_8class_eval.csv` ← **`54_english_balanced_8class_eval.ipynb`** (по желанию)
- Base eval slice: `notebooks/results/english_dataset_alignment/32_english_base_eval_slice.csv`
- English city-swap пары: `notebooks/results/english_dataset_alignment/32_english_city_swap_counterfactuals.csv` ← **32**

**Чекпоинты:** см. `notebooks/models/english/checkpoint_resolution.json`; опционально symlinks в `notebooks/models/english/<model_run_id>/`.

**Выходы (без коллизий между моделями):**

- `notebooks/results/english_eval/<model_run_id>/` и `figures/english_eval/<model_run_id>/`
- `notebooks/results/english_city_swap_eval/<model_run_id>/` после **e05**

Отсутствующие CSV (например soft slice) **пропускаются** с пометкой в `run_manifest.json`.

**Важно:** корневой `notebooks/60_label_smoothing.ipynb` — это **другое** (обучение), не путать с **e04**.


## Порядок запуска (рекомендуемый)

1. **`32_english_dataset_alignment.ipynb`** — сплиты, base slice, **city-swap counterfactuals** (обязательно до e01/e05 по данным).
2. **`54_english_balanced_8class_eval.ipynb`** (по желанию) — если нужен soft slice в метриках.
3. **`59_english_transfer_existing_models.ipynb`** (`notebooks/`, не эта папка) — если нужна отдельная папка `english_transfer_eval/` для OOD-формулировки; можно **параллельно** с e01–e03.
4. **e01, e02, e03** — по одной модели; **можно параллельно**, если хватает GPU/RAM.
5. **e05** — после появления чекпоинтов и файла `32_english_city_swap_counterfactuals.csv`; **можно параллельно** с e01–e03.
6. **e04** — **после** того как в `english_eval/` (и при необходимости `english_transfer_eval/`) лежат актуальные `metrics.json`.
7. Опционально: **`52_table_export.ipynb`** — таблица **T15** из `english_benchmark_summary.csv`; **`71`** если нужен join с русским треком в сводке e04.
